# Global Supply Chain Analytics

## End-to-End Data Analytics Project

### Project Overview

This project presents a comprehensive analysis of a global retail company's supply chain operations using transactional data. The objective is to evaluate business performance across sales, profitability, customer behavior, product performance, and logistics.

The analysis follows a structured data analytics workflow beginning with business understanding, followed by data quality assessment, data cleaning, feature engineering, exploratory data analysis, and business reporting.

The final deliverables include an interactive Power BI dashboard, business recommendations, and executive-level insights to support data-driven decision-making.

# Business Understanding

## Background

DataCo Global is a multinational retail company that operates across multiple countries and markets, serving customers through a large-scale supply chain network.

Over time, the organization has observed several operational challenges, including inconsistent profitability, delayed deliveries, varying customer purchasing patterns, and uneven sales performance across different regions and product categories.

To better understand these issues, the management has requested a detailed analysis of historical transactional data.

This project aims to identify operational inefficiencies, evaluate business performance, and provide actionable recommendations that can improve profitability, customer satisfaction, and overall supply chain efficiency.

# Business Objectives

The primary objective of this project is to evaluate the company's supply chain performance and generate actionable business insights.

The analysis aims to:

- Evaluate overall sales and profitability.
- Identify high-performing and underperforming products.
- Analyze customer purchasing behavior across different customer segments.
- Assess shipping performance and delivery efficiency.
- Measure the impact of discounts on business performance.
- Compare operational performance across countries, markets, and regions.
- Identify opportunities to improve profitability and customer satisfaction.
- Support strategic business decision-making through data-driven insights.

# Observation

The columns **Sales per customer** and **Order Item Total** contain identical values for all 180,519 records, indicating that they represent the same business metric.

Additionally, it was verified that:

### **Sales − Order Item Discount = Order Item Total**

This confirms that:

- **Sales** represents the gross sales amount before discount.
- **Order Item Discount** represents the discount applied.
- **Order Item Total** represents the final amount paid by the customer after discount.

Since **Sales per customer** is an exact duplicate of **Order Item Total**, it is removed from the dataset to eliminate redundancy.

# Observation

### It was verified that **Benefit per order** and **Order Profit Per Order** contain identical values across all records.

Since both columns represent the same business metric, **Benefit per order** was removed to eliminate redundancy, while **Order Profit Per Order** was retained due to its clearer and more descriptive naming.

# KPI Definition Sheet

The following KPI definitions establish a standardized measurement framework for the project. Each KPI has a clearly defined business meaning, source column, and calculation logic to ensure consistency across Python analysis and the Power BI dashboard.

| KPI | Business Definition | Source Column(s) | Calculation Logic |
|---|---|---|---|
| Gross Revenue | Total sales value before discounts | Sales | SUM(Sales) |
| Net Revenue | Revenue generated after discounts | Order Item Total | SUM(Order Item Total) |
| Total Profit | Total profit generated from all order items | Order Profit Per Order | SUM(Order Profit Per Order) |
| Total Orders | Number of unique orders | Order Id | DISTINCTCOUNT(Order Id) |
| Total Customers | Number of unique customers | Customer Id | DISTINCTCOUNT(Customer Id) |
| Total Quantity Sold | Total number of units sold | Order Item Quantity | SUM(Order Item Quantity) |
| Average Order Value | Average net revenue generated per order | Net Revenue, Order Id | Net Revenue / Total Orders |
| Average Delivery Delay | Average difference between actual and scheduled shipping time | Shipping Delay | AVERAGE(Shipping Delay) |
| Late Delivery Rate | Percentage of orders shipped later than scheduled | Delivery Performance, Order Id | Delayed Orders / Total Orders |
| Overall Profit Margin | Percentage of net revenue retained as profit | Net Revenue, Total Profit | Total Profit / Net Revenue × 100 |
